```python
df = df.sort_values("date")
```

```python
df = df.set_index("date")
```

```python
df = df.reset_index()
```

```python
# 按索引筛选出 2024-01-01 到 2024-03-01 之间的数据行。
df.loc["2024-01-01":"2024-03-01"]
```

```python
# ret = close/lastClose - 1  pct_change就是这个意思
df["ret"]=df["close"].pct_change()
```

```python
df["ret"]
```

```python
# diff() 的意思是：计算当前值和上一行值的差值。
df["diff"]=df["close"].diff()
df["diff"]
```

```python
df["amount"] = df["close"] * df["volume"]
df["amount"]
```

```python
# 统计每一列有多少个缺失值
df.isna().sum()
```

```python
df = df.dropna()
df.isna().sum()
```

```python
# 填充缺失值为0
df = df.fillna(0)
# 用前一个值填充
df = df.ffill()
```

```python
# df["strategy_ret"] = df["symbol"] * df["ret"]
# shift 是什么含义 shift(1)   # 取上一行 shift(-1)  # 取下一行
df["strategy_ret"] = df["signal"].shift(1) * df["ret"]
df["strategy_ret"]
```

```python
df["ma20"]=df["close"].rolling(20).mean()
df["ma20"]
```

```python
df["ma60"]=df["close"].rolling(60).mean()
df["ma60"]
```

```python
df["vol20"]=df["ret"].rolling(20).std()
df["vol20"]
```

```python
# cumprod() 的含义是累计连乘
df["equity"]=(1+df["ret"]).cumprod()
df["equity"]
```

```python
# 按照哪一列来分组
df["ret"] = df.groupby("symbol")["close"].pct_change()
df["equity"] = (1 + df["ret"]).groupby(df["symbol"]).cumprod()
df[["symbol", "date", "close", "ret", "equity"]]
```

```python
# transform的意思是  输入：每个 symbol 分组后的 close  输出：和原 df 一样长的一列结果
df["ma20"] = df.groupby("symbol")["close"].transform(lambda x: x.rolling(20).mean())
df[["ma20"]]
```

```python
# agg() 的核心意思是：把多行数据压缩成统计结果。
summary = df.groupby("symbol").agg({
    "close":"mean",
    "volume":"sum",
    "amount":"sum"
    })
display(summary)
```

```python
df.groupby("symbol")["close"].agg(["mean", "max", "min", "std"])
```
