```python
import pandas as pd


def calculate_cagr(start_value, end_value, years):
    """
    计算 CAGR 年化收益率
    """
    if start_value <= 0:
        raise ValueError("start_value must be > 0")
    if end_value <= 0:
        raise ValueError("end_value must be > 0")
    if years <= 0:
        raise ValueError("years must be > 0")

    return (end_value / start_value) ** (1 / years) - 1


df = pd.read_csv("../data/sample_etf_daily.csv")
df["date"]=pd.to_datetime(df["date"])
df = df.sort_values(["symbol", "date"])

for symbol,group in df.groupby("symbol"):
    # 这句是对当前这一组数据做清洗
    group = group.dropna(subset=["close"])
    if len(group)<2:
        continue
    years = len(group)/250
    startPrice = group["close"].iloc[0]
    endPrice = group["close"].iloc[-1]

    cagr=calculate_cagr(startPrice,endPrice,years)
    results = []
    results.append({
            "symbol": symbol,
            "years": years,
            "cagr": cagr,
            "cagr_percent": f"{cagr:.2%}"
        })
    print(results)
```
